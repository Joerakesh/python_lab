## Aim
Use pandas groupby aggregations, custom coefficient-of-variation aggregation, pivot tables and group-wise transform for revenue analysis.

## Algorithm
1. Create a sales DataFrame with region, product, quantity and revenue.
2. Group by region and calculate total revenue, mean revenue and coefficient of variation.
3. Create a Region × Product pivot table.
4. Use groupby().transform() to compute each row's share of regional revenue.
5. Display summaries and transformed row-level values.

## Notes
The reference summary below is transcribed from the supplied practical record where legible. Running this notebook produces fresh outputs from the code above.

**Recorded reference result:** The practical record reports regional total revenues East=4000, South=3100 and West=4150. The row-level regional revenue shares are printed in the record; executing the notebook regenerates the complete summary tables.

```python
# Data Aggregation and Group-wise Operations
# Course: Python for Data Science | Loyola College, Chennai
# Student: Joe Rakesh A | Department No.: 26-PDS-002

# Exercise 13 — Data Aggregation and Group-wise Operations
import pandas as pd

pd.set_option("display.width", 120)
sales = pd.DataFrame({
    "Region": ["East", "West", "East", "South", "West", "South", "East", "West"],
    "Product": ["A", "B", "A", "C", "A", "B", "C", "C"],
    "Quantity": [10, 15, 8, 20, 12, 5, 18, 9],
    "Revenue": [1000, 1800, 800, 2500, 1400, 600, 2200, 950],
})

agg_result = sales.groupby("Region").agg(
    Total_Revenue=("Revenue", "sum"),
    Avg_Revenue=("Revenue", "mean"),
    Revenue_CV=("Revenue", lambda values: values.std() / values.mean() if values.mean() else 0),
)
print("Custom aggregation (sum, mean, coefficient of variation) by Region:\n", agg_result)

pivot = pd.pivot_table(
    sales, values="Revenue", index="Region", columns="Product", aggfunc="sum", fill_value=0
)
print("\nPivot table (Revenue by Region x Product):\n", pivot)

sales["Revenue_Share_in_Region"] = sales.groupby("Region")["Revenue"].transform(
    lambda values: values / values.sum()
)
print("\nRow-level share of regional revenue (via transform):\n",
      sales[["Region", "Product", "Revenue", "Revenue_Share_in_Region"]])
```
